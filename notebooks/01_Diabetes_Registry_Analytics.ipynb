🚀 Phase 8.1 — Statistical Assumption Checking

Before running statistical tests, a professional analyst checks assumptions.

Research Question 1

Is HbA1c normally distributed?

```python
from scipy.stats import shapiro

# Use a sample because 700,000 records is too large for Shapiro-Wilk
sample = df["HbA1c_pct"].sample(5000, random_state=42)

stat, p = shapiro(sample)

print("Shapiro-Wilk Statistic:", round(stat,4))
print("P-value:", round(p,4))
```

# Phase 8.2 — Independent t-test
*Research Question*

Do male and female patients differ in their mean HbA1c?

```python
from scipy.stats import ttest_ind

male = df[df["Sex"] == "Male"]["HbA1c_pct"]
female = df[df["Sex"] == "Female"]["HbA1c_pct"]

t_stat, p_value = ttest_ind(male, female)

print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.5f}")
```

# Phase 8.3 — Effect Size (Very Important)

*Statistical significance alone isn't enough, especially with 700,000 patients. Even tiny differences can produce very small p-values.*

Let's calculate Cohen's d, which measures the magnitude of the difference.

```python
import numpy as np

mean_diff = male.mean() - female.mean()

pooled_sd = np.sqrt(
    ((male.std()**2) + (female.std()**2)) / 2
)

cohens_d = mean_diff / pooled_sd

print(f"Cohen's d = {cohens_d:.3f}")
```

## Phase 9: Predictive Analytics (Machine Learning)

*This is the section that will make your GitHub project stand out.*

```python
## Step 1 — Create Target Variable
# Convert Yes/No to 1/0
df["HighRisk"] = df["High_Risk"].map({
    "Yes":1,
    "No":0
})

df["HighRisk"].value_counts()

## Step 2 — Select Features
# We will start with clinically meaningful variables.

features = [
    "Age",
    "BMI",
    "HbA1c_pct",
    "Creatinine_mg_dL",
    "eGFR",
    "FBS_mmol_L",
    "RBS_mmol_L",
    "LDL_mg_dL"
]

X = df[features]

y = df["HighRisk"]


## Step 3 — Train-Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Check the sizes:

print(X_train.shape)
print(X_test.shape)


## Step 4 — Build Your First Model
# We start with Logistic Regression.

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)


## Step 5 — Make Predictions
predictions = model.predict(X_test)

predictions[:20]


## Step 6 — Evaluate the Model
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", round(accuracy,3))


## Step 7 — Confusion Matrix
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, predictions)

cm


## Step 8 — Plot Confusion Matrix
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.title("Confusion Matrix")

plt.show()


## Step 9 — Classification Report
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    predictions
))


## Step 10 — Feature Importance
#For logistic regression, inspect the coefficients.

importance = pd.DataFrame({
    "Feature":features,
    "Coefficient":model.coef_[0]
})

importance.sort_values(
    by="Coefficient",
    ascending=False
)
```
